````python
# Import required libraries
import pandas as pd
````

## Step 1: Load the Dataset from Parquet

Load the dataset from `gold_dataset.parquet` using `pandas`.

````python
# Load the dataset from Parquet
file_path = 'artifacts/ingestions/datas/gold_dataset.parquet'
df = pd.read_parquet(file_path)

# Display basic info
print('Dataset shape:', df.shape)
print('\nFirst 5 rows:')
display(df.head())
````

## Step 2: Inspect Column Names and Data Types

Verify the columns for **machine** (`machine_id_std`) and **time** (`window_start`).

````python
# Check column names and data types
print('Columns:', df.columns.tolist())
print('\nData types:')
print(df.dtypes)
````

## Step 3: Sort by Machine and Time

Sort the DataFrame by `machine_id_std` (ascending) and `window_start` (ascending).

````python
# Sort the DataFrame
df_sorted = df.sort_values(by=['machine_id_std', 'window_start'])

# Verify the sorted data
print('First 15 rows after sorting:')
display(df_sorted.head(15))
````

## Step 4: Verify Sorting for a Specific Machine

Check if the data is correctly sorted by machine and time.

````python
# Check sorting for a specific machine
sample_machine = df_sorted['machine_id_std'].iloc[0]
machine_data = df_sorted[df_sorted['machine_id_std'] == sample_machine]

print(f'Sample machine: {sample_machine}')
print(f'Number of records for {sample_machine}: {len(machine_data)}')
print('\nFirst 5 records for this machine:')
display(machine_data.head())
````

## Step 5: Export Sorted Data (Optional)

Save the sorted DataFrame to a new Parquet file.

````python
# Export to Parquet
output_parquet_path = 'artifacts/ingestions/datas/gold_dataset_sorted.parquet'
df_sorted.to_parquet(output_parquet_path, index=False)
print(f'Sorted data exported to: {output_parquet_path}')
````

## Notes
- **Leakage Columns**: Any column containing future information (e.g., `future_incident_count_*`, `label_failure_next_*`) must be excluded to avoid data leakage.
- **Identifiers**: Columns like `machine_id_std`, `window_start`, and `window_end` are not useful as features.
- **Temporal Split**: The `split_set` column provides a predefined temporal split (train < validation < test). Do not use random splitting.
- **Imputation**: Median imputation is used for numerical features to handle missing values.
- **Standardization**: Required for linear models (e.g., logistic regression, SVM).
- **Horizon Selection**: Choose the horizon (e.g., `label_failure_next_24h`) based on the predictive maintenance task requirements.

````python
# Define columns to exclude
exclude_columns = [
    # Identifiers
    'machine_id_std', 'window_start', 'window_end',
    # Leakage columns (future information)
    'future_incident_count_6h', 'label_failure_next_6h',
    'future_incident_count_12h', 'label_failure_next_12h',
    'future_incident_count_24h', 'label_failure_next_24h',
    'future_incident_count_48h', 'label_failure_next_48h',
    # Incident counts (potential leakage)
    'incident_count_1h', 'incident_max_severity_1h',
    'incident_count_prev_24h', 'incident_max_severity_prev_24h',
    'incident_count_prev_7d', 'hours_since_last_incident',
    # Type-specific incident counts (potential leakage)
    'type_surchauffe_count_prev_24h', 'type_baisse_pression_count_prev_24h',
    'type_vibration_count_prev_24h', 'type_bruit_mecanique_count_prev_24h',
    'type_surconsommation_count_prev_24h', 'type_blocage_mecanique_count_prev_24h',
    'type_alarme_capteur_count_prev_24h', 'type_arret_urgence_count_prev_24h',
    'type_defaut_qualite_count_prev_24h',
    # Maintenance columns (potential leakage)
    'days_since_last_maintenance', 'maintenance_count_prev_30d',
    # Split column (to be used later)
    'split_set'
]

# Exclude columns from the DataFrame
features_df = df.drop(columns=exclude_columns)

# Verify remaining columns
print("Remaining columns:", features_df.columns.tolist())
````

````python
# Select the horizon (e.g., 24h)
horizon = 'label_failure_next_48h'

# Extract the target variable
y = df[horizon].astype(int)  # Convert boolean to 0/1

# Verify target distribution
print("Target distribution:")
print(y.value_counts())

# Split the data using the 'split_set' column
train_df = df[df['split_set'] == 'train']
val_df = df[df['split_set'] == 'validation']
test_df = df[df['split_set'] == 'test']

# Split features and target for each set
X_train = train_df.drop(columns=exclude_columns + [horizon])
y_train = train_df[horizon].astype(int)

X_val = val_df.drop(columns=exclude_columns + [horizon])
y_val = val_df[horizon].astype(int)

X_test = test_df.drop(columns=exclude_columns + [horizon])
y_test = test_df[horizon].astype(int)

# Verify shapes
print(f"Train shape: {X_train.shape}, {y_train.shape}")
print(f"Validation shape: {X_val.shape}, {y_val.shape}")
print(f"Test shape: {X_test.shape}, {y_test.shape}")
````

````python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Create a pipeline for imputation and standardization
pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),  # Impute NaN with median
    ('scaler', StandardScaler())  # Standardize features
])

# Fit and transform the training data
X_train_processed = pipeline.fit_transform(X_train)

# Transform validation and test data (using the same pipeline)
X_val_processed = pipeline.transform(X_val)
X_test_processed = pipeline.transform(X_test)

# Verify no NaN values remain
print("NaN values in X_train:", pd.isna(X_train_processed).sum().sum())
print("NaN values in X_val:", pd.isna(X_val_processed).sum().sum())
print("NaN values in X_test:", pd.isna(X_test_processed).sum().sum())
````
